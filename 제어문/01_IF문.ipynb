```python
# IF문은 왜 필요할까?
"돈이 있으면 택시를 타고 가고, 돈이 없으면 걸어간다"
```

```python
money = True

if money : print("택시를 타고가라.")
else : print("걸어가라")
```

```python
# if문의 기본 구조

# if 조건문:
#     수행 할 문장1
#     수행 할 문장2
#     ...
# else:
#     수행 할 문장1
#     수행 할 문장2
#     ...
```

```python
# # 들여쓰기 방법
# if 조건문:
#     수행 할 문장1
# 수행 할 문장2 -> if문 안에서 실행이 되지 않음 이후로 에러 발생
#     수행 할 문장3 -> 에러발생

money = True
if money:
    print("택시를")
    print("타고")
    print("가라")
```

```python
#만약 주머니에 money가 있으면 택시를 타고가고 그렇지 않으면 걸어가라
pocket = ["paper", "phone", "money"]

if "money" in pocket:
    print("택시를 타고가라") 

else:
    print("걸어가라")
```

```python
# 다양한 조건을 판단하는 elif
pocket = ["paper", "phone"]
card =True

if "money" in pocket:
    print("택시를 타고 가라")
else:
    if card == True:
        print("택시를 타고가라")
    else:
        print("걸어가라")
        
```

```python
# 조건부 표현식
score = 85

if score >= 60:
    result = "합격"
else:
    result = "불합격"
print(result)
```

```python
#학점이 90 이상이면 A, 80이상 90 미만이면 B, 70이상 80미만이면 C, 60이상 70미만이면 D , 60미만이면 F

score = input("학점을 입력하세요 : ")
if   score >= 90:
    print(f"입력값은 {score}, 당신의 학점은 : A")
elif score >= 80:
    print(f"입력값은 {score}, 당신의 학점은 : B")
elif score >= 70:
    print(f"입력값은 {score}, 당신의 학점은 : C")
elif score >= 60:
    print(f"입력값은 {score}, 당신의 학점은 : D")
else:
    print(f"입력값은 {score}, 당신의 학점은 : F")
```
